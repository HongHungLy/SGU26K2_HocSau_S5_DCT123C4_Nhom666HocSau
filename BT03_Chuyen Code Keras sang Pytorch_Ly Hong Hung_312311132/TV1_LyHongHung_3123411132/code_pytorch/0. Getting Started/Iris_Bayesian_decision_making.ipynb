# (1) import libraries

```python
from sklearn import datasets
from sklearn.naive_bayes import GaussianNB

import numpy as np
import time

# Cố định seed để kết quả chia tập train/test có tính tái lập
np.random.seed(42)
```

# (2) preparing iris dataset

```python
iris = datasets.load_iris()

# Lấy ngẫu nhiên 60% chỉ số cho tập huấn luyện và 40% còn lại cho tập kiểm tra
n_samples = iris.target.shape[0]
train_size = int(n_samples * 0.6)
train_indx = np.random.choice(range(0, n_samples), train_size, replace=False)
test_indx = np.array([indx for indx in np.arange(0, n_samples, 1) if indx not in train_indx])

x_train = iris.data[train_indx, :]
y_train = iris.target[train_indx]

x_test = iris.data[test_indx, :]
y_test = iris.target[test_indx]

print("x_train shape:", x_train.shape)
print("y_train shape:", y_train.shape)
print("x_test shape :", x_test.shape)
print("y_test shape :", y_test.shape)
```

# (3) parameters estimation

```python
# Ước lượng vector kỳ vọng (mean) và ma trận hiệp phương sai (covariance) cho từng lớp (0, 1, 2)
for cls in range(3):
    mu = np.mean(x_train[y_train == cls], axis=0)
    sigma = np.cov(x_train[y_train == cls], rowvar=False)
    print(f"Class {cls} - mean:\n{mu}")
    print(f"Class {cls} - covariance:\n{sigma}\n")
```

# (4) bayesian decision making

```python
gnb = GaussianNB()

start_time = time.time()
y_pred = gnb.fit(x_train, y_train).predict(x_test)
elapsed = time.time() - start_time

acc = (y_test == y_pred).sum() / x_test.shape[0]
print('Testing classification accuracy: %.4f (time: %.6f s)' % (acc, elapsed))
```
